```python
import os

# 2. ВКАЖІТЬ ІМ'Я ВАШОГО ЗАВАНТАЖЕНОГО ФАЙЛУ ТУТ:
INPUT_PDF = "testyvan.pdf"
OUTPUT_DIR = "./output"

if not os.path.exists(INPUT_PDF):
    print(f"ПОМИЛКА: Файл {INPUT_PDF} не знайдено! Завантажте його у панелі зліва.")
else:
    # Запускаємо Marker
    # (Під час першого запуску він буде завантажувати ~3 ГБ моделей, це нормально)
    !marker_single $INPUT_PDF $OUTPUT_DIR --batch_multiplier 2 --max_pages 500
    
    print(f"\n\n=== ГОТОВО! ===")
    print(f"Перейдіть у папку {OUTPUT_DIR} зліва, там буде папка з вашим файлом.")
    print(f"Ви можете завантажити .md файл і зображення назад на ваш комп'ютер.")
```
